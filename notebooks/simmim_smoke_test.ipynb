# Initial Mask Ratio Comparison

We ran a small single-GPU smoke-test experiment using the official Swin-based SimMIM configuration on a tiny ImageNet-style dataset, while changing only the mask ratio and keeping other settings fixed.

Settings:
- Config: `simmim_pretrain__swin_base__img192_window6__100ep.yaml`
- Image size: 192
- Batch size: 2
- Epochs: 1
- GPU: Tesla T4
- Data: tiny ImageNet-style sample set

Results:

| Mask Ratio | Loss |
|---|---:|
| 0.25 | 0.9771 |
| 0.50 | 1.0081 |
| 0.75 | 1.0458 |

Observation:
As the mask ratio increases, the training loss also increases. This suggests that heavier masking makes the reconstruction task more difficult, which is consistent with the intuition behind masked image modeling.

## Note

These results are only from a tiny smoke-test setting with a very small dataset and one training epoch, so they should be interpreted as an initial trend rather than a final quantitative conclusion.
